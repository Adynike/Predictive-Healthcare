#### Calculate Outliers using Empirical Rule and z-score
** The what, why and the when

##### Empirical Rule (What, why and when)
- What it is:
Assumes data is normally distributed.
Anything outside mean ± 3 standard deviations is considered an outlier.

- When to use it:
When data is roughly bell-shaped (normal distribution).
When you want a quick, simple outlier check.
When the dataset is large and the distribution is stable

- Why use it:
Fast and easy.
Good for variables like heights, lab values, many biological measures.
Works well if data follows a normal pattern.

##### Z-Score Method (What, why and when)
- What it is:
Measures how many standard deviations a value is from the mean.

- When to use it:
When there is need for more precision than the empirical rule.
When the distribution is approximately normal but not perfect.
When comparing multiple variables on different scales.
To score how extreme each point is, not just flag it.

- Why use it:
Works even if the distribution is not perfectly normal.
Gives exact "extremeness" of each value (useful in modeling).
More flexible (threshold can be adjusted)



#### Calculate Outlier using Empirical Rule

```python
# Drop NaN values and include percentage (%) of the outliers

for name, data in list_data:
    print(f"\n=== OUTLIER SUMMARY: {name} ===")

    for col in data.select_dtypes(include='number'):
        col_data = data[col].dropna()
        
        mean = col_data.mean()
        std = col_data.std()
        
        lower = mean - 3*std
        upper = mean + 3*std
        
        outliers = data[(data[col] < lower) | (data[col] > upper)]
        
        percent = (len(outliers) / len(data)) * 100
        
        print(f"{col}: {len(outliers)} outliers ({percent:.2f}%)")
    
    print('_' * 70)
```

#### Calculate Ouliers using Z-score for evry numeric column in each dataset

```python
for name, data in list_data:
    print(f"\n=== Z-SCORE OUTLIER SUMMARY: {name} ===")

    numeric_cols = data.select_dtypes(include='number').columns

    for col in numeric_cols:
        col_data = data[col].dropna()
        mean = col_data.mean()
        std = col_data.std()

        if std == 0:
            print(f"{col}: STD = 0 → Skipped")
            print('-' * 50)
            continue

        z = (col_data - mean) / std
        outliers = col_data[abs(z) > 3]

        print(f"Column: {col}")
        print(f"Outlier Count: {len(outliers)}")
        print('-' * 50)

    print('_' * 70)
```

## Task 1
Check the incorrect spellings and convert to null values and handle appropriately

```python
for name, data in list_data:
    print(f"Cleaning {name}...")

    # Ward code typos
    if 'ward_code' in data.columns:
        data['ward_code'] = data['ward_code'].replace({'SURCG': np.nan})

    # Sex column rare/invalid entries
    if 'sex' in data.columns:
        data['sex'] = data['sex'].replace({'X': np.nan})

    # Outcome rare/invalid
    if 'outcome' in data.columns:
        data['outcome'] = data['outcome'].replace({'unknown': np.nan})

    # Dates: convert invalid to NaT
    if 'date' in data.columns:
        data['date'] = pd.to_datetime(data['date'], errors='coerce')

    # triage_level: convert invalid strings to NaN
    if 'triage_level' in data.columns:
        data['triage_level'] = pd.to_numeric(data['triage_level'], errors='coerce')

    print(f"{name} cleaned.\n")
```

#### Handle missing values after cleaning

```python
for name, data in list_data:
    print(f"Handling missing values in {name}...")

    if 'triage_level' in data.columns and 'ward_code' in data.columns:
        data['triage_level'] = data.groupby('ward_code')['triage_level'].transform(
            lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else np.nan)
        )

    # Categorical columns filled with mode
    for col in ['ward_code', 'sex', 'outcome', 'arrival_source']:
        if col in data.columns and data[col].isna().sum() > 0:
            data[col] = data[col].fillna(data[col].mode()[0])

    # Fill missing dates using forward-fill, then backward-fill
    if 'date' in data.columns:
        data['date'] = data['date'].fillna(method='ffill').fillna(method='bfill')

    print(f"Missing values handled for {name}.\n")
```

```python
# Check for remaining missing values

for name, data in list_data:
    print(f"{name} missing values summary:\n")
    print(data.isna().sum())
    print("-" * 50)
```

```python
for name, data in list_data:
    print(f"Handling remaining missing values in {name}...")

    # Fill triage_level by ward mode, then overall mode
    if 'triage_level' in data.columns and 'ward_code' in data.columns:
        data['triage_level'] = data.groupby('ward_code')['triage_level'].transform(
            lambda x: x.fillna(x.mode().iloc[0] if not x.mode().empty else np.nan)
        )
        data['triage_level'] = data['triage_level'].fillna(data['triage_level'].mode()[0])

    # Fill categorical columns with mode
    for col in ['ward_code', 'sex', 'outcome', 'arrival_source']:
        if col in data.columns:
            data[col] = data[col].fillna(data[col].mode()[0])

    # Fill missing dates
    if 'date' in data.columns:
        data['date'] = data['date'].fillna(method='ffill').fillna(method='bfill')

   
    print(f"Missing values handled for {name}.\n")
```

##### Reasons why merging immediately is not an option
We do not merge immediately because the datasets are at different granularities:
1) admissions is patient-level (one row per admission)
2) metrics is daily/ward-level (one row per ward per day).

If merge immediately:
- We might end up duplicating metric rows for multiple admissions per day.
- Any analysis could be misleading, e.g., inflating bed occupancy or admission rates.

##### What to be done before merging
1) Aggregate the admissions data so it is at the same level as metrics: daily + hospital + ward.
2) Engineer meaningful features from admissions to summarize patient-level data.

##### How to merge
Once admissions are aggregated to daily + hospital + ward, merge with metrics using key columns like: date, hospital_id, ward_code.
- This ensures a one-to-one or many-to-one match between admissions and metrics.
